Importing Model

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Importing Datasets

```python
dataset = pd.read_csv('Position_Salaries.csv')
X = dataset.iloc[:, 1:-1].values
y = dataset.iloc[:, -1].values
```

```python
X
```

```python
y
```

Training Decision Tree Regression model on whole dataset

```python
from sklearn.tree import DecisionTreeRegressor

dtr_regressior = DecisionTreeRegressor(random_state=0)
dtr_regressior.fit(X, y)
```

Predicting a new result

```python
y_predict = dtr_regressior.predict([[5.99]])
y_predict
```

Visualising the Decision Tree Regression Results

```python
plt.scatter(X, y, color='red')
plt.plot(X, dtr_regressior.predict(X), color='blue')
plt.title('Salary Expectation')
plt.xlabel('Level of employee')
plt.ylabel('Salary')
plt.show()
```

Visualising the Decision Tree Regression Results(higher Resolution)

```python
X_grid = np.arange(min(X), max(X), 0.1)
X_grid = X_grid.reshape((-1, 1))
plt.scatter(X, y, color='red')
plt.plot(X_grid, dtr_regressior.predict(X_grid), color='blue')
plt.title('Salary Expectation')
plt.xlabel('Level of employee')
plt.ylabel('Salary')
plt.show()
```
